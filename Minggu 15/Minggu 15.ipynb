Latihan 15.1

```python
def prima(n, p=2):
    if n < 2:
        return "Bukan Bilangan Prima!"

    if n % 2 == 0:
        return "Bukan Bilangan Prima!"

    if p * p > n:
        return "Bilangan Prima!"

    if n % p == 0:
        return "Bukan Bilangan Prima!"

    return prima(n, p + 1)

bilangan = int(input("Masukkan bilangan : "))
print(prima(bilangan))
```

Latihan 15.2

```python
def palindrom(kal):
    left = 0
    right = len(kal) - 1
    if kal[left] == kal[right]:
        return True
    elif kal[left] != kal[right]:
        return False
    return palindrom(kal, left + 1, right - 1)

kal = input("Masukkan Kalimat : ")
if palindrom(kal.lower().replace(" ", "")):
    print("Palindrom!")
else:
    print("Bukan Palindrom!")
```

Latihan 15.3

```python
def sumDeret(n):
    if n == 1:
        return 1
    else:
        return sumDeret(n - 1) + (2**n - 1)
    
n = int(input("Batas : "))
print(sumDeret(n))
```

Latihan 15.4

```python
def digit(angka):
    if angka < 10:
        return angka
    return (angka % 10) + digit(angka // 10)

angka = int(input("Bilangan : "))
print(digit(angka))
```

Latihan 15.5

```python
def kombine(n, r):
    if r == 0 or r == n:
        return 1
    
    return kombine(n - 1, r - 1) + kombine(n - 1, r)

n = int(input("Bil1 : "))
r = int(input("Bil2 : "))
if r > n:
    print("r Harus lebih kecil dari n")
else:
    print(f"C({n}, {r}) = {kombine(n, r)}")
```
