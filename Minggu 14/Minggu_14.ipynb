Latihan 14.1

```python
n = int(input('Masukkan jumlah kategori: '))
data_aplikasi = {}

for i in range(n):
    nama_kategori = input('Masukkan nama kategori: ')
    print('Masukkan 5 nama aplikasi di kategori', nama_kategori)

    aplikasi = []
    for j in range(5):
        nama_aplikasi = input('Nama aplikasi: ')
        aplikasi.append(nama_aplikasi)

    data_aplikasi[nama_kategori] = aplikasi

print(data_aplikasi)

daftar_aplikasi_list = []

for aplikasi in data_aplikasi.values():
    daftar_aplikasi_list.append(set(aplikasi))
print(f"List Aplikasi: {daftar_aplikasi_list}")

semua = daftar_aplikasi_list[0]
for i in range(1, len(daftar_aplikasi_list)):
    semua = semua.intersection(daftar_aplikasi_list[i])

print(f"Aplikasi Yang Ada Di semua Kategori: {semua}")

aplik = set()
for i in range(len(daftar_aplikasi_list)):
    for j in range(i+1, len(daftar_aplikasi_list)):
        aplik = aplik.union(daftar_aplikasi_list[i].intersection(daftar_aplikasi_list[j]))

semua_aplikasi = daftar_aplikasi_list[0]
for i in range(1, len(daftar_aplikasi_list)):
    semua_aplikasi = semua_aplikasi.union(daftar_aplikasi_list[i])

satu = semua_aplikasi - aplik

print(f"Aplikasi Yang Hanya Ada Di satu Kategori: {satu}")

if n > 2:
    dua = aplik - semua
    print(f"Aplikasi Yang Hanya Ada Di Dua Kategori Sekaligus: {dua}")
```

```python
data_list = [10, 12, 20, 40, 60]
list_set = set(data_list)
print(f"List Awal: {data_list}")
print(f"List Menjadi Set: {list_set}\n")

data_set = {10, 12, 20, 40, 60}
set_list = list(data_set)
print(f"Set Awal: {data_set}")
print(f"Set Menjadi List: {set_list}\n")

data_tuple = (10, 12, 20, 40, 60)
tuple_set = set(data_tuple)
print(f"Tuple Awal: {data_tuple}")
print(f"Tuple Menjadi Set: {tuple_set}\n")

data_set2 = {10, 12, 20, 40, 60}
set_tuple = tuple(data_set2)
print(f"Set Awal: {data_set2}")
print(f"Set Menjadi Tuple: {set_tuple}")
```

```python
import string
file1 = input("File 1 : ")
file2 = input("File 2 : ")

try:
    fop1 = open(file1)
    fop2 = open(file2)

    kata1 = list()
    kata2 = list()

    for i in fop1:
        kata = i.translate(str.maketrans("","",string.punctuation)).lower().strip("\n")
        for j in kata.split():
            kata1.append(j)
    for k in fop2:
        kata = k.translate(str.maketrans("","",string.punctuation)).lower().strip("\n")
        for l in kata.split():
            kata2.append(l)

    print(f"Kata di File 1: {kata1}")
    print(f"Jumlah Kata : {len(kata1)} kata")
    print(f"Kata di File 2: {kata2}")
    print(f"Jumlah Kata : {len(kata2)} kata")
    print(f"")
    print(f"Jumlah Kata Unik: \nFile1 : {len(set(kata1))} kata \nFile2 : {len(set(kata2))} kata\n")

    sama = set(kata1) & set(kata2)
    print(f"Kata Yang muncul pada kedua file: {sama}")
    print(f"Jumlah Kata : {len(sama)} kata")

except:
    print("File tidak ditemukan!")
```
