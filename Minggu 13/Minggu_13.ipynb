```python
def dataaa(data):
    nim = data[1]
    nama = data[0]
    alamat =  data[2]

    print(f"NIM     : {nim}")
    print(f"Nama    : {nama}")
    print(f"Alamat  : {alamat}\n")

    depan = nama.split()
    nama_depan = depan[0]
    print(f"NIM: {tuple(nim)}\n")
    print(f"NAMA DEPAN: {tuple(nama_depan[1:])}\n")
    print(f"NAMA TERBALIK: {tuple(depan[::-1])}")

data = ("Mikael Marlisvemba Lejo Saputro", "71251179", "Marau, Ketapang")
dataaa(data)
```

Latihan 13.3

```python
import re
fname = input("Enter a file name: ")
fopen = open(fname)


jam_count = {}

for baris in fopen:
    if not baris.startswith("From "):
        continue
    bagian = re.findall(r"(\d{2}):\d{2}:\d{2}", baris)
    jam = bagian[0]
    if jam not in jam_count:
        jam_count[jam] = 1
    else:
       jam_count[jam] += 1 

for jam in sorted(jam_count):
    print(jam, jam_count[jam])
```
