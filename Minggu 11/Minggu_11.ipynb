Latihan 10.1

```python
def terbaik(hello):
    hello_int = [int(i) for i in hello]
    hello_int.sort()
    hello_int.reverse()
    if len(hello_int) >= 3:
        print(f"Bilangan Terbaiik : {hello_int[:3]}")
    else:
        print("Minimal 3 Bilangan di dalam List!")

hello = input().split()
terbaik(hello)
```

```python
def rata_rata():
    list_angka = []
    rata2 = 0

    while True:
        angka = input("Masukkan Input : ")
        if angka == "done":
            jumlah = sum(list_angka)
            rata2 = jumlah/len(list_angka)
            print(f"Program Selesai!\nRata-Rata : {rata2}")
            break
        else:
            list_angka.append(int(angka))

rata_rata()
```

```python
import re

def kata_unik(file):
    with open(file) as f:
        kkata = f.read()

    kata = re.findall(r"[a-zA-Z]+", kkata.lower())

    unik = sorted(set(kata))

    for i in range(1, len(unik)+1):
        print(f" {i}. {unik[i-1]}")

file = input("Masukkan Nama File : ")

kata_unik(file)
```
