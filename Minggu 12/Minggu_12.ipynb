```python
def dictionary(dictio):
    item = 0
    print("Key     Values    Item")
    for key, values in dictio.items():
        item += 1
        print(f"{key}       {values}        {item}")

dictio = dict(eval(input()))
dictionary(dictio)
```

```python
def gabung(list1, list2):
    dictio = {}
    if len(list1) == len(list2):
        for i in range(max(len(list1), len(list2))):
            dictio.update({list1[i] : list2[i]})
        print(dictio)
    else:
        print("List Harus Sama Panjang!")
    
list1 = list(eval(input()))
list2 = list(eval(input()))
gabung(list1, list2)
```

Latihan 12.3

```python
import re

def cariEmail(file):
    mail = {}
    for line in file:
        email = re.findall(r"From (\b\S+@\S+\b)", line)
        for i in email:
            mail.update({i : mail.get(i, 0) + 1})

    print(mail)
 
fname = input("Nama File : ")
file = open(fname)
cariEmail(file)
```
