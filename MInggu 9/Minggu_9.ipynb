Latihan 8.1

```python
def anagram(txt, ana):
    if sorted(txt) == sorted(ana):
        return True
    else:
        return False
    
text = input("Masukkan Kata Awal : ")
ana = input("Masukkan Kata yang Ingin di Cek : ")

if anagram(text.lower(), ana.lower()):
    print(f"'{ana}' Merupakan Anagram Dari '{text}'")
else:
    print(f"'{ana}' Bukan Merupakan Anagram Dari '{text}'")
```

Latihan 8.2

```python
import re

def makan(text, kata):
    txt = re.findall(kata, text)
    jumlah = txt.count(kata)
    return(jumlah)
    
    
text = input("Masukkan Input : ")
kata = input("Masukan Kata Yang Ingin Dicari Jumlahnya : ")
ls = text.lower()
ms = kata.lower()
hasil = makan(ls, ms)
print(f"Output: {kata} ada {hasil} buah")
```

Latihan 8.3

```python
import re

def spasi(txt):
    hasil = re.sub(r"\s+", " ", txt)
    return hasil.strip()

txt = input("Masukkan text : ")
print(f"Teks Asli :'{txt}'")
print(f"Hasil : '{spasi(txt)}'")
```

Latihan 8.4

```python
import re

def pjpd(txt):
    test = txt.split()
    text = re.findall(r"[a-zA-Z]+", str(test))
    pendek = min(text, key=len)
    panjang = max(text, key=len)
    terpendek = ', '.join([i for i in text if len(i) == len(pendek)])
    terpanjang = ', '.join([j for j in text if len(j) == len(panjang)])
    print(f"Terpendek : {terpendek} \nTerpanjang : {terpanjang}")

play = input("Masukkan Teks : ")
pjpd(play)
```

Latihan 8.5

```python
import re
from datetime import datetime

def tanggal(txt):
    lst = re.findall(r"\d{4}-\d{2}-\d{2}",txt)
    for tgl_str in lst:
        tgl = datetime.strptime(tgl_str, "%Y-%m-%d")
        baru = datetime.strftime(tgl, "%d-%m-%Y")
        hari_ini = datetime.today()
        selisih = hari_ini - tgl
        jumlah = selisih.days
        print(f"{tgl} atau {baru}, selisih {jumlah} hari")

txt = input("Masukkan Text : ")
tanggal(txt)
```

```python
import re
import random
import string

def mail(txt):
    karakter = string.ascii_letters + string.digits
    lst = re.findall(r"[a-z]+@[a-z]+.(?:co\.id|\.com)",txt)
    user = re.findall(r"([a-z]+)@", str(lst))
    for i in range(len(lst)):
        password = ''.join(random.choices(karakter, k=8))
        print(f"{lst[i]}, username : {user[i]}, password : {password}")

txt = input("Masukkan Teks : ")
mail(txt)
```
