```python
def bandingFile(satu, dua):
    try:
        with open(satu, 'r') as s, open(dua, 'r') as d:
            baris_s = s.readlines()
            baris_d = d.readlines()

        beda = False
        baris = max(len(baris_s), len(baris_d))

        if baris == 0:
            print("====== File Kosong ======")
        else:
            for i in range(baris):
                bs = baris_s[i].rstrip("\n") if i < len(baris_s) else "Baris Tidak Ada!"
                bd = baris_d[i].rstrip("\n") if i < len(baris_d) else "Baris Tidak Ada!"

                if bs != bd:
                    print("Terdapat Perbedaan!!")
                    beda = True
                    print(f"Baris Ke-{i+1}")
                    print(f"{satu} = {bs}")
                    print(f"{dua} = {bd}\n")

            if not beda:
                print("File Sama!!")
            else:
                print("File Berbeda!!")

            print(f"Jumlah Baris Yang di Cek : {baris}")
    except:
        print("File Tidak Ditemukan")


satu = input("Nama File Pertama : ")
dua = input("Nama File Kedua : ")
bandingFile(satu, dua)
```

```python
def jawabSoal(file):
    try:
        with open(file) as soal:
            tanya = soal.readlines()
 
        baris = len(tanya)

        for i in range(baris):
            pertanyaan = tanya[i].rstrip("\n").split("||")
            soalan = pertanyaan[0]
            jawaban = pertanyaan[1].lower().strip()

            print(soalan)
            jawab = input("Jawab : ")
            
            if jawab.lower() == jawaban:
                print("Jawaban Benar!!")
            else:
                print("Jawaban Salah!!")
    
    except:
        print("File Tidak Ditemukan!!")


file = input("Nama File : ")
jawabSoal(file)
```
